```python
class Node:
  def __init__(self, data, next= None):
    self.__data = data
    self.__next = next

  @property
  def data(self):
    return self.__data

  @property
  def next(self):
    return self.__next

  @data.setter
  def data(self, data):
    self.__data = data

  @next.setter
  def next(self, nodo):
    self.__next = nodo
```

```python
class Lista:
  def __init__(self, head=None, tail=None, size=0):
    self.__head = None
    self.__tail = None
    self.__size = 0

  @property
  def size(self):
    return self.__size

  @size.setter
  def size(self,s):
    self.__size = s

  def isEmpty(self):
    return self.__size==0

  def First(self):
    return self.__head

  def Last(self):
    return self.__tail

  def addFirst(self, e):
    Nodo = Node(e, None)

    if self.isEmpty():
      self.__head = Nodo
      self.__tail = Nodo
    else:
      Nodo.next = self.__head
      self.__head = Nodo
    self.__size += 1

  def addLast(self, e):
    Nodo = Node(e, None)

    if self.isEmpty():
      self.__head = Nodo
      self.__tail = Nodo
    else:
      self.__tail.next = Nodo
      self.__tail = Nodo
    self.__size += 1

  def removeFirst(self):
    if not self.isEmpty():
      temp = self.__head
      self.__head = temp.next
      temp.next = None
      self.__size -= 1
      if self.__size == 0:
                self.__tail = None
      return temp.data
    else:
      return None

  def removeLast(self):
    if (self.__size == 1):
      return self.removeFirst()

    elif (self.__size > 1):
      temp = self.__tail
      anterior = self.__head

      while (anterior.next != self.__tail):
        anterior = anterior.next
      anterior.next = None
      self.__tail = anterior
      self.__size -= 1
      return temp.data

    else:
      return None
```

```python
class DoubleNodo:

  def __init__(self, data, next = None, prev = None):
    self.__data = data
    self.__next = next
    self.__prev = prev

  def getData(self):
    return self.__data

  def getNext(self):
    return self.__next

  def getPrev(self):
    return self.__prev

  def setData(self, d):
    self.__data = d

  def setNext(self, n):
    self.__next = n

  def setPrev(self, p):
    self.__prev = p
```

```python
class DoubleList:

  def __init__(self):
    self.__head = None
    self.__tail = None
    self.__size = 0

  def size(self):
    return self.__size

  def isEmpty(self):
    return self.__size == 0

  def first(self):
    return self.__head

  def last(self):
    return self.__tail

  def addFirst(self, e):
    nodo = DoubleNodo(e)

    if self.isEmpty():
      self.__head = nodo
      self.__tail = nodo

    else:
      nodo.setNext(self.__head)
      self.__head.setPrev(nodo)
      self.__head = nodo
    self.__size += 1

  def addLast(self, e):
        nodo = DoubleNodo(e)
        if self.isEmpty():
            self.__head = nodo
            self.__tail = nodo
        else:
            self.__tail.setNext(nodo)
            nodo.setPrev(self.__tail)
            self.__tail = nodo
        self.__size += 1

  def removeFirst(self):
        if not self.isEmpty():
            temp_dato = self.__head.getData()
            if self.__size == 1:
                self.__head = None
                self.__tail = None
            else:
                temp = self.__head.getNext()
                self.__head.setNext(None)
                temp.setPrev(None)
                self.__head = temp
            self.__size -= 1
            return temp_dato
        else:
            return None


  def removeLast(self):
    if not self.isEmpty():
      temp_dato = self.__tail.getData()
      if self.__size == 1:
        self.__head = None
        self.__tail = None
      else:
        temp = self.__tail.getPrev()
        self.__tail.setPrev(None)
        temp.setNext(None)
        self.__tail = temp
        self.__size -= 1
      return temp_dato
    else:
      return None

  def remove(self, n):
    if n == self.__head:
      return self.removeFirst()
    elif n == self.__tail:
      return self.removeLast()
    else:
      temp_data = n.getData()
      temp_prev = n.getPrev()
      temp_next = n.getNext()
      temp_prev.setNext(temp_next)
      temp_next.setPrev(temp_prev)
      n.setNext(None)
      n.setPrev(None)
      self.__size -= 1
      return temp_data

  def addAfter(self, n, e):
    if n == self.__tail:
      self.addLast(e)
    else:
      m = DoubleNodo(e)
      temp = n.getNext()
      n.setNext(m)
      m.setPrev(n)
      m.setNext(temp)
      temp.setPrev(m)
      self.__size +=1

  def addBefore(self, n, e):
    if n == self.__head:
      self.addFirst(e)
    else:
      m = DoubleNodo(e)
      temp = n.getPrev()
      temp.setNext(m)
      m.setPrev(temp)
      m.setNext(n)
      n.setPrev(m)
      self.__size +=1
```

```python
class Pila:
    def __init__(self):
        self.__datos = Lista()

    def isEmpty(self):
        return self.__datos.isEmpty()

    def size(self):
        return self.__datos.size

    def push(self, elemento):
        self.__datos.addFirst(elemento)

    def pop(self):
        return self.__datos.removeFirst()

    def top(self):
        if not self.isEmpty():
            return self.__datos.First().data
        return None


class Cola:
    def __init__(self):
        self.__datos = Lista()

    def isEmpty(self):
        return self.__datos.isEmpty()

    def size(self):
        return self.__datos.size

    def enqueue(self, elemento):
        self.__datos.addLast(elemento)

    def dequeue(self):
        return self.__datos.removeFirst()

    def first(self):
        if not self.isEmpty():
            return self.__datos.First().data
        return None
```

```python
class TreeNode:
    def __init__(self, key, data):
        self.__key = key
        self.__data = data
        self.__left = None
        self.__right = None

    @property
    def key(self):
        return self.__key

    @property
    def data(self):
        return self.__data

    @data.setter
    def data(self, data):
        self.__data = data

    @property
    def left(self):
        return self.__left

    @left.setter
    def left(self, node):
        self.__left = node

    @property
    def right(self):
        return self.__right

    @right.setter
    def right(self, node):
        self.__right = node
```
