```python
class Node:
  def __init__(self, data, next= None):
    self.__data = data
    self.__next = next

  @property
  def data(self):
    return self.__data

  @property
  def next(self):
    return self.__next

  @data.setter
  def data(self, data):
    self.__data = data

  @next.setter
  def next(self, nodo):
    self.__next = nodo
```

```python
class Lista:
  def __init__(self, head=None, tail=None, size=0):
    self.__head = None
    self.__tail = None
    self.__size = 0

  @property
  def size(self):
    return self.__size

  @size.setter
  def size(self,s):
    self.__size = s

  def isEmpty(self):
    return self.__size==0

  def First(self):
    return self.__head

  def Last(self):
    return self.__tail

  def addFirst(self, e):
    Nodo = Node(e, None)

    if self.isEmpty():
      self.__head = Nodo
      self.__tail = Nodo
    else:
      Nodo.next = self.__head
      self.__head = Nodo
    self.__size += 1

  def addLast(self, e):
    Nodo = Node(e, None)

    if self.isEmpty():
      self.__head = Nodo
      self.__tail = Nodo
    else:
      self.__tail.next = Nodo
      self.__tail = Nodo
    self.__size += 1

  def removeFirst(self):
    if not self.isEmpty():
      temp = self.__head
      self.__head = temp.next
      temp.next = None
      self.__size -= 1
      if self.__size == 0:
                self.__tail = None
      return temp.data
    else:
      return None

  def removeLast(self):
    if (self.__size == 1):
      return self.removeFirst()

    elif (self.__size > 1):
      temp = self.__tail
      anterior = self.__head

      while (anterior.next != self.__tail):
        anterior = anterior.next
      anterior.next = None
      self.__tail = anterior
      self.__size -= 1
      return temp.data

    else:
      return None
```

```python
class DoubleNodo:

  def __init__(self, data, next = None, prev = None):
    self.__data = data
    self.__next = next
    self.__prev = prev

  def getData(self):
    return self.__data

  def getNext(self):
    return self.__next

  def getPrev(self):
    return self.__prev

  def setData(self, d):
    self.__data = d

  def setNext(self, n):
    self.__next = n

  def setPrev(self, p):
    self.__prev = p
```

```python
class DoubleList:

  def __init__(self):
    self.__head = None
    self.__tail = None
    self.__size = 0

  def size(self):
    return self.__size

  def isEmpty(self):
    return self.__size == 0

  def first(self):
    return self.__head

  def last(self):
    return self.__tail

  def addFirst(self, e):
    nodo = DoubleNodo(e)

    if self.isEmpty():
      self.__head = nodo
      self.__tail = nodo

    else:
      nodo.setNext(self.__head)
      self.__head.setPrev(nodo)
      self.__head = nodo
    self.__size += 1

  def addLast(self, e):
        nodo = DoubleNodo(e)
        if self.isEmpty():
            self.__head = nodo
            self.__tail = nodo
        else:
            self.__tail.setNext(nodo)
            nodo.setPrev(self.__tail)
            self.__tail = nodo
        self.__size += 1

  def removeFirst(self):
        if not self.isEmpty():
            temp_dato = self.__head.getData()
            if self.__size == 1:
                self.__head = None
                self.__tail = None
            else:
                temp = self.__head.getNext()
                self.__head.setNext(None)
                temp.setPrev(None)
                self.__head = temp
            self.__size -= 1
            return temp_dato
        else:
            return None


  def removeLast(self):
    if not self.isEmpty():
      temp_dato = self.__tail.getData()
      if self.__size == 1:
        self.__head = None
        self.__tail = None
      else:
        temp = self.__tail.getPrev()
        self.__tail.setPrev(None)
        temp.setNext(None)
        self.__tail = temp
        self.__size -= 1
      return temp_dato
    else:
      return None

  def remove(self, n):
    if n == self.__head:
      return self.removeFirst()
    elif n == self.__tail:
      return self.removeLast()
    else:
      temp_data = n.getData()
      temp_prev = n.getPrev()
      temp_next = n.getNext()
      temp_prev.setNext(temp_next)
      temp_next.setPrev(temp_prev)
      n.setNext(None)
      n.setPrev(None)
      self.__size -= 1
      return temp_data

  def addAfter(self, n, e):
    if n == self.__tail:
      self.addLast(e)
    else:
      m = DoubleNodo(e)
      temp = n.getNext()
      n.setNext(m)
      m.setPrev(n)
      m.setNext(temp)
      temp.setPrev(m)
      self.__size +=1

  def addBefore(self, n, e):
    if n == self.__head:
      self.addFirst(e)
    else:
      m = DoubleNodo(e)
      temp = n.getPrev()
      temp.setNext(m)
      m.setPrev(temp)
      m.setNext(n)
      n.setPrev(m)
      self.__size +=1
```

```python
class Pila:
    def __init__(self):
        self.__datos = Lista()

    def isEmpty(self):
        return self.__datos.isEmpty()

    def size(self):
        return self.__datos.size

    def push(self, elemento):
        self.__datos.addFirst(elemento)

    def pop(self):
        return self.__datos.removeFirst()

    def top(self):
        if not self.isEmpty():
            return self.__datos.First().data
        return None


class Cola:
    def __init__(self):
        self.__datos = Lista()

    def isEmpty(self):
        return self.__datos.isEmpty()

    def size(self):
        return self.__datos.size

    def enqueue(self, elemento):
        self.__datos.addLast(elemento)

    def dequeue(self):
        return self.__datos.removeFirst()

    def first(self):
        if not self.isEmpty():
            return self.__datos.First().data
        return None
```

```python
class TreeNode:
    def __init__(self, key, data):
        self.__key = key
        self.__data = data
        self.__left = None
        self.__right = None

    @property
    def key(self):
        return self.__key

    @property
    def data(self):
        return self.__data

    @data.setter
    def data(self, data):
        self.__data = data

    @property
    def left(self):
        return self.__left

    @left.setter
    def left(self, node):
        self.__left = node

    @property
    def right(self):
        return self.__right

    @right.setter
    def right(self, node):
        self.__right = node
```

```python
class BinarySearchTree:
    def __init__(self):
        self.__root = None
        self.__size = 0

    @property
    def size(self):
        return self.__size

    def isEmpty(self):
        return self.__root is None

    def Insert(self, key, data):
        if self.isEmpty():
            self.__root = TreeNode(key, data)
            self.__size += 1
        else:
            self.insert_recursive(self.__root, key, data)

    def insert_recursive(self, current, key, data):
        if key < current.key:
            if current.left is None:
                current.left = TreeNode(key, data)
                self.__size += 1
            else:
                self.insert_recursive(current.left, key, data)
        elif key > current.key:
            if current.right is None:
                current.right = TreeNode(key, data)
                self.__size += 1
            else:
                self.insert_recursive(current.right, key, data)
        else:
            current.data = data

    def search(self, key):
        return self.search_recursive(self.__root, key)

    def search_recursive(self, current, key):
        if current is None:
            return None
        if key == current.key:
            return current.data
        elif key < current.key:
            return self.search_recursive(current.left, key)
        else:
            return self.search_recursive(current.right, key)

    def print_in_order(self):
        self.in_order_recursive(self.__root)

    def in_order_recursive(self, current):
        if current is not None:
            self.in_order_recursive(current.left)
            print(f"[Clave: {current.key}] -> {current.data}")
            self.in_order_recursive(current.right)
```

```python
class Usuario:
  def __init__(self, id, nombre, edad, telefono, password=""):
    self.__id = int(id)
    self.__nombre = nombre
    self.__edad = int(edad)
    self.__telefono = telefono
    self.__password = password

  @property
  def id(self):
    return self.__id

  @property
  def nombre(self):
    return self.__nombre

  @nombre.setter
  def nombre(self, nombre):
    self.__nombre = nombre

  @property
  def edad(self):
    return self.__edad

  @edad.setter
  def edad(self, edad):
    self.__edad = edad

  @property
  def telefono(self):
    return self.__telefono

  @telefono.setter
  def telefono(self, telefono):
    self.__telefono = telefono

  def validar_password(self, clave):
    return self.__password == clave

  def __str__(self):
    return f"ID: {self.__id}\nNombre: {self.__nombre}\nEdad: {self.__edad}\nTelefono: {self.__telefono}"

```

```python
class Deportista(Usuario):
  def __init__(self, id, nombre, edad, telefono, deporte, peso, categoria, experienciaMeses):
    super().__init__(id, nombre, edad, telefono)
    self.__deporte = deporte
    self.__peso = float(peso)
    self.__categoria = categoria
    self.__experienciaMeses = experienciaMeses
    self.__historial_lesiones = Pila()
    self.__historial_pagos = Pila()

  @property
  def deporte(self):
    return self.__deporte

  @property
  def peso(self):
    return self.__peso

  @peso.setter
  def peso(self, peso):
    self.__peso = peso

  @property
  def categoria(self):
    return self.__categoria

  @categoria.setter
  def categoria(self, categoria):
    self.__categoria = categoria

  @property
  def experienciaMeses(self):
    return self.__experienciaMeses

  @experienciaMeses.setter
  def experienciaMeses(self, experiencia):
    self.__experienciaMeses = experiencia


  def registrar_lesion(self, fecha, descripcion, gravedad):
    registro = f"{fecha} | {descripcion} (Gravedad: {gravedad})"
    self.__historial_lesiones.push(registro)

  def ultima_lesion(self):
    return self.__historial_lesiones.top() or "Sin lesiones registradas"

  def registrar_pago(self, mes, monto, fecha):
    recibo = f"Mes: {mes} | ${monto} | Pagado el {fecha}"
    self.__historial_pagos.push(recibo)

  def ultimo_pago(self):
    return self.__historial_pagos.top() or "Sin pagos registrados"

  def __str__(self):
    return (f"{super().__str__()} | {self.__deporte} - Cat: {self.__categoria} Peso: {self.__peso}kg | Exp: {self.__experienciaMeses} meses")
```

```python
class Deporte:
    def __init__(self, nombre_disciplina):
        self.__nombre_disciplina = nombre_disciplina
        self.__roster = DoubleList()
        self.__indice_torneo = BinarySearchTree()
        self.__cola_asistencia = Cola()

    @property
    def nombre_disciplina(self):
        return self.__nombre_disciplina

    @property
    def roster(self):
        return self.__roster

    @property
    def indice_torneo(self):
        return self.__indice_torneo

    @property
    def cola_asistencia(self):
        return self.__cola_asistencia

    def inscribir_deportista(self, deportista):
        if self.__indice_torneo.search(deportista.id) is not None:
            return False
        self.__roster.addLast(deportista)
        self.__indice_torneo.Insert(deportista.id, deportista)
        return True

    def buscar_deportista(self, id_deportista):
        return self.__indice_torneo.search(int(id_deportista))

    def registrar_llegada(self, id_deportista):
        deportista = self.buscar_deportista(id_deportista)
        if deportista:
            self.__cola_asistencia.enqueue(deportista)
            return True
        return False

    def procesar_siguiente_asistencia(self):
        return self.__cola_asistencia.dequeue()
```

```python
class Entrenador(Usuario):
    def __init__(self, id, nombre, edad, telefono, password, especialidad):
        super().__init__(id, nombre, edad, telefono, password)
        self.__especialidad = especialidad

    @property
    def especialidad(self):
        return self.__especialidad

    def registrar_nuevo_deportista(self, club_deporte, id_dep, nombre, edad, tel, peso, cat, exp):
        nuevo = Deportista(id_dep, nombre, edad, tel, club_deporte.nombre_disciplina, peso, cat, exp)
        return club_deporte.inscribir_deportista(nuevo)

    def consultar_ficha_torneo(self, club_deporte, id_dep):
        dep = club_deporte.buscar_deportista(id_dep)
        if not dep:
            return None
        return {
            "ficha": str(dep),
            "ultimo_pago": dep.ultimo_pago(),
            "ultima_lesion": dep.ultima_lesion()
        }

    def modificar_datos_deportista(self, club_deporte, id_dep, nuevo_peso=None, nueva_cat=None, nueva_exp=None):
        dep = club_deporte.buscar_deportista(id_dep)
        if not dep:
            return False
        if nuevo_peso is not None:
            dep.peso = nuevo_peso
        if nueva_cat is not None:
            dep.categoria = nueva_cat
        if nueva_exp is not None:
            dep.experienciaMeses = nueva_exp
        return True
```

```python
import gradio as gr
# --- 1. INICIALIZACIÓN EN MEMORIA ---
disciplinas = {
    "Boxeo": Deporte("Boxeo"),
    "Baloncesto": Deporte("Baloncesto"),
    "Patinaje": Deporte("Patinaje")
}
admin = Entrenador(9001, "Carlos Mendoza", 34, "3009876543", "admin123", "Director Técnico")

# Datos precargados en memoria para demostrar funcionamiento inmediato
admin.registrar_nuevo_deportista(disciplinas["Boxeo"], 1001, "Mateo Ríos", 18, "3001112233", 64.5, "Welter", 14)
admin.registrar_nuevo_deportista(disciplinas["Boxeo"], 1005, "Valeria Gómez", 19, "3004445566", 57.0, "Pluma", 20)
admin.registrar_nuevo_deportista(disciplinas["Boxeo"], 1003, "Andrés Castro", 17, "3007778899", 71.0, "Mediano", 8)

mateo = disciplinas["Boxeo"].buscar_deportista(1001)
mateo.registrar_pago("Octubre 2026", 80000, "2026-10-05")
mateo.registrar_pago("Noviembre 2026", 80000, "2026-11-03")
mateo.registrar_lesion("2026-09-12", "Esguince leve muñeca", "Leve")


# --- 2. FUNCIONES DE CONEXIÓN CON ESTRUCTURAS ---
def _recorrer_in_orden(nodo, filas):
    if nodo is not None:
        _recorrer_in_orden(nodo.left, filas)
        dep = nodo.data
        filas.append([
            dep.id, dep.nombre, dep.edad, f"{dep.peso} kg",
            dep.categoria, f"{dep.experienciaMeses} m",
            dep.ultimo_pago(), dep.ultima_lesion()
        ])
        _recorrer_in_orden(nodo.right, filas)


def listar_roster(disciplina_nombre):
    club = disciplinas[disciplina_nombre]
    filas = []
    _recorrer_in_orden(club.indice_torneo._BinarySearchTree__root, filas)
    return filas


def registrar_deportista_ui(disciplina_nombre, id_dep, nombre, edad, tel, peso, cat, exp):
    try:
        club = disciplinas[disciplina_nombre]
        exito = admin.registrar_nuevo_deportista(
            club, int(id_dep), nombre, int(edad), tel, float(peso), cat, int(exp)
        )
        if exito:
            return f"✅ {nombre} inscrito en {disciplina_nombre}.", listar_roster(disciplina_nombre)
        return "⚠️ Ya existe un deportista con ese ID.", listar_roster(disciplina_nombre)
    except Exception as e:
        return f"❌ Error en los datos: {e}", listar_roster(disciplina_nombre)


def modificar_deportista_ui(disciplina_nombre, id_dep, peso, cat, exp):
    try:
        club = disciplinas[disciplina_nombre]
        nuevo_peso = float(peso) if peso not in (None, "", 0) else None
        nueva_cat = cat.strip() if cat and cat.strip() != "" else None
        nueva_exp = int(exp) if exp not in (None, "", 0) else None

        if admin.modificar_datos_deportista(club, int(id_dep), nuevo_peso, nueva_cat, nueva_exp):
            return f"✅ Datos del ID {int(id_dep)} actualizados.", listar_roster(disciplina_nombre)
        return "❌ Deportista no encontrado.", listar_roster(disciplina_nombre)
    except Exception as e:
        return f"❌ Error al actualizar: {e}", listar_roster(disciplina_nombre)


def consultar_torneo_ui(disciplina_nombre, id_dep):
    try:
        club = disciplinas[disciplina_nombre]
        info = admin.consultar_ficha_torneo(club, int(id_dep))
        if not info:
            return "❌ Deportista no encontrado en el Árbol Binario."
        return (f"📋 FICHA TÉCNICA:\n{info['ficha']}\n\n"
                f"💰 ÚLTIMO PAGO (Tope Pila):\n{info['ultimo_pago']}\n\n"
                f"🩹 ÚLTIMA LESIÓN (Tope Pila):\n{info['ultima_lesion']}")
    except Exception:
        return "❌ Ingresa un ID numérico válido."


def registrar_historial_ui(disciplina_nombre, id_dep, tipo, campo1, campo2, campo3):
    try:
        club = disciplinas[disciplina_nombre]
        dep = club.buscar_deportista(int(id_dep))
        if not dep:
            return "❌ Deportista no encontrado.", listar_roster(disciplina_nombre)

        if tipo == "Pago Mensualidad":
            dep.registrar_pago(campo1, campo2, campo3)
            msg = f"💰 Pago apilado en el historial de {dep.nombre}."
        else:
            dep.registrar_lesion(campo1, campo2, campo3)
            msg = f"🩹 Lesión apilada en el historial de {dep.nombre}."

        return msg, listar_roster(disciplina_nombre)
    except Exception as e:
        return f"❌ Error: {e}", listar_roster(disciplina_nombre)


def gestionar_asistencia_ui(disciplina_nombre, accion, id_dep):
    club = disciplinas[disciplina_nombre]
    if accion == "Registrar Llegada (Encolar FIFO)":
        try:
            if club.registrar_llegada(int(id_dep)):
                return f"➡️ ID {int(id_dep)} entró a la Cola. En espera: {club.cola_asistencia.size()} atletas."
            return "❌ ID no registrado en esta disciplina."
        except Exception:
            return "❌ Ingresa un ID numérico."
    else:
        atleta = club.procesar_siguiente_asistencia()
        if atleta:
            return f"📣 Turno de asistencia: {atleta.nombre} (ID: {atleta.id}). Quedan en cola: {club.cola_asistencia.size()}."
        return "ℹ️ La Cola de asistencia está vacía."


# --- 3. INTERFAZ VISUAL ---
with gr.Blocks(theme=gr.themes.Soft(), title="Club Saetas Valientes") as app:
    gr.Markdown("# 🏹 Gestión Deportiva — Club Saetas Valientes")
    gr.Markdown("Implementación en memoria con `DoubleList`, `BinarySearchTree`, `Pila` y `Cola`.")

    disciplina_selector = gr.Radio(
        choices=["Boxeo", "Baloncesto", "Patinaje"],
        value="Boxeo",
        label="Disciplina Activa"
    )

    with gr.Tabs():
        with gr.Tab("📋 Roster Ordenado (Árbol In-Orden)"):
            btn_refrescar = gr.Button("🔄 Actualizar Tabla")
            tabla_roster = gr.Dataframe(
                headers=["ID", "Nombre", "Edad", "Peso", "Categoría", "Exp.", "Último Pago (Pila)", "Última Lesión (Pila)"],
                value=listar_roster("Boxeo"),
                interactive=False
            )
            btn_refrescar.click(listar_roster, inputs=disciplina_selector, outputs=tabla_roster)
            disciplina_selector.change(listar_roster, inputs=disciplina_selector, outputs=tabla_roster)

        with gr.Tab("➕ Inscribir / Modificar Deportista"):
            with gr.Row():
                with gr.Column():
                    gr.Markdown("### Nuevo Ingreso")
                    in_id = gr.Number(label="ID / Documento", precision=0)
                    in_nombre = gr.Textbox(label="Nombre Completo")
                    in_edad = gr.Number(label="Edad", precision=0)
                    in_tel = gr.Textbox(label="Teléfono")
                    in_peso = gr.Number(label="Peso (kg)")
                    in_cat = gr.Textbox(label="Categoría")
                    in_exp = gr.Number(label="Experiencia (meses)", precision=0)
                    btn_guardar = gr.Button("Inscribir (DoubleList + Árbol)", variant="primary")
                with gr.Column():
                    gr.Markdown("### Actualizar Ficha Física")
                    mod_id = gr.Number(label="ID del Deportista", precision=0)
                    mod_peso = gr.Number(label="Nuevo Peso (dejar en 0 para no cambiar)")
                    mod_cat = gr.Textbox(label="Nueva Categoría (vacío para no cambiar)")
                    mod_exp = gr.Number(label="Nueva Experiencia meses (0 para no cambiar)", precision=0)
                    btn_modificar = gr.Button("Actualizar Datos")
                    out_crud = gr.Textbox(label="Resultado", interactive=False)

            btn_guardar.click(
                registrar_deportista_ui,
                inputs=[disciplina_selector, in_id, in_nombre, in_edad, in_tel, in_peso, in_cat, in_exp],
                outputs=[out_crud, tabla_roster]
            )
            btn_modificar.click(
                modificar_deportista_ui,
                inputs=[disciplina_selector, mod_id, mod_peso, mod_cat, mod_exp],
                outputs=[out_crud, tabla_roster]
            )

        with gr.Tab("🏆 Torneo e Historial (Pilas LIFO)"):
            with gr.Row():
                with gr.Column():
                    q_id = gr.Number(label="Buscar ID en Árbol O(log n)", precision=0)
                    btn_buscar = gr.Button("Consultar Ficha Rápida", variant="primary")
                    out_ficha = gr.Textbox(label="Ficha y Tope de Pilas", lines=8, interactive=False)
                with gr.Column():
                    gr.Markdown("### Apilar Nuevo Registro (Push)")
                    h_id = gr.Number(label="ID del Deportista", precision=0)
                    h_tipo = gr.Radio(["Pago Mensualidad", "Lesión Clínica"], value="Pago Mensualidad", label="Tipo")
                    h_c1 = gr.Textbox(label="Mes (Pago) o Fecha YYYY-MM-DD (Lesión)")
                    h_c2 = gr.Textbox(label="Monto $ (Pago) o Descripción (Lesión)")
                    h_c3 = gr.Textbox(label="Fecha Pago (Pago) o Gravedad (Lesión)")
                    btn_apilar = gr.Button("Apilar Registro (Push)")
                    out_historial = gr.Textbox(label="Confirmación", interactive=False)

            btn_buscar.click(consultar_torneo_ui, inputs=[disciplina_selector, q_id], outputs=out_ficha)
            btn_apilar.click(
                registrar_historial_ui,
                inputs=[disciplina_selector, h_id, h_tipo, h_c1, h_c2, h_c3],
                outputs=[out_historial, tabla_roster]
            )

        with gr.Tab("⏱️ Asistencia a Clase (Cola FIFO)"):
            a_accion = gr.Radio(
                ["Registrar Llegada (Encolar FIFO)", "Llamar Siguiente Atleta (Desencolar FIFO)"],
                value="Registrar Llegada (Encolar FIFO)",
                label="Acción en Cola"
            )
            a_id = gr.Number(label="ID del Deportista (para Encolar)", precision=0)
            btn_cola = gr.Button("Ejecutar en Cola", variant="primary")
            out_cola = gr.Textbox(label="Estado de la Fila", interactive=False)

            btn_cola.click(gestionar_asistencia_ui, inputs=[disciplina_selector, a_accion, a_id], outputs=out_cola)

if __name__ == "__main__":
    app.launch()
```
